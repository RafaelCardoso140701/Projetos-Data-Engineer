# [ITEM 3]
## Usando o _BeautifulSoup4_ e o _pandas_, extraia a tabela de UFs brasileiras da [página da Wikipédia de UFs brasileiras](https://pt.wikipedia.org/wiki/Unidades_federativas_do_Brasil), que está na secção [Lista sintética](https://colab.research.google.com/corgiredirector?site=https%3A%2F%2Fpt.wikipedia.org%2Fwiki%2FUnidades_federativas_do_Brasil%23Lista_sint%25C3%25A9tica), da página, e monte um dataframe com as informações.

### O dataframe montado deve seguir os requisitos abaixo:

*   ### O dataframe deve conter todas as UFs brasileiras.
*   ### O dataframe só precisa ter as seguintes colunas:
###`['Bandeira', 'UnidadeFederativa', 'Abreviacao', 'Populacao', 'PIB']`.
*   ### No caso da coluna `'Bandeira'`, usar a URL da imagem como valor.
*   ### No caso das colunas `'Populacao'` e `'PIB'`, deixar os valores em `int` no dataframe.
*   ### Dê `.strip()` nas strings antes de colocar no dataframe.

````python
import pandas as pd
import requests
from bs4 import BeautifulSoup

url = 'https://pt.wikipedia.org/wiki/Unidades_federativas_do_Brasil'

headers = {
    "User-Agent": "Mozilla/5.0"
}

response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text, 'html.parser')

# pega todas as tabelas
tables = soup.find_all('table')

table = None

# encontra a tabela correta pelo cabeçalho REAL
for t in tables:
    if "População" in t.text and "PIB" in t.text:
        table = t
        break

if table is None:
    raise ValueError("Tabela não encontrada.")

rows = table.find_all('tr')

data = []

def clean_numeric(value):
    return int(
        value.replace('.', '')
             .replace(',', '')
             .replace('\xa0', '')
             .strip()
    )

for row in rows[1:]:
    cols = row.find_all('td')

    if len(cols) < 8:
        continue

    try:
        # Bandeira
        img = cols[0].find('img')
        bandeira = img['src'] if img else None
        if bandeira and not bandeira.startswith('http'):
            bandeira = 'https:' + bandeira

        # Dados
        uf = cols[1].text.strip()
        abrev = cols[2].text.strip()
        populacao = clean_numeric(cols[5].text.strip())
        pib = clean_numeric(cols[7].text.strip())

        data.append({
            'Bandeira': bandeira,
            'UnidadeFederativa': uf,
            'Abreviacao': abrev,
            'Populacao': populacao,
            'PIB': pib
        })

    except:
        continue

df = pd.DataFrame(data)

print(df.head())
print(f"\nTotal de UFs: {len(df)}")
````

# [ITEM 4]
### Filtre o dataframe do __ITEM 3__ para obter as linhas que correspondem às UFs com população maior ou igual a 10 milhões.

````python
# Considerando que o df do ITEM 3 já foi criado

# Filtrar UFs com população >= 10 milhões
df_filtrado = df[df['Populacao'] >= 10_000_000].copy()

# (opcional) ordenar da maior para menor população
df_filtrado = df_filtrado.sort_values(by='Populacao', ascending=False)

# resetar índice
df_filtrado.reset_index(drop=True, inplace=True)

print(df_filtrado)
print(f"\nTotal de UFs com população >= 10 milhões: {len(df_filtrado)}")
````
